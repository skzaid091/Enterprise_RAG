```python
@dataclass  
class PageContent:
    page_number: int    # 1-based
    text: str
    word_count: int
```

```python
@dataclass
class PDFDocument:
    file_path: str
    page_count: int
    pages: list[PageContent]   # one per page
    metadata: dict             # title, author, etc from PDF headers
    error: str | None          # None if success
```

```python
class PDFLoader:
    
    def load(self, pdf_path):

        if not os.path.exists(pdf_path):
            return PDFDocument(
                file_path=pdf_path,
                page_count=0,
                pages=[],
                metadata={},
                error=f"File not found: {pdf_path}"
            )
        
        try:
            doc = pymupdf.open(pdf_path)
            
        except Exception as e:
            return PDFDocument(
                file_path=pdf_path,
                page_count=0,
                pages=[],
                metadata={},
                error=f"Failed to open PDF: {e}"
            )

        # metadata
        raw = doc.metadata
        metadata = {
            "title":   raw.get("title"),
            "author":  raw.get("author"),
            "subject": raw.get("subject"),
        }

        pages = []
        for page in doc:
            text = page.get_text().strip()
            
            pages.append(PageContent(
                page_number=page.number + 1,
                text=text,
                word_count=len(text.split())
            ))

        return PDFDocument(
            file_path=pdf_path,
            page_count=len(doc),
            pages=pages,
            metadata=metadata,
            error=None
        )


    def load_documents(self, documents):
        processed_documents = []
        
        for document_path in documents:
            processed_documents.append(
                self.load(document_path)
            )

        return processed_documents
```

```python
@dataclass
class Chunk:
    chunk_id: int

    document_id: int
    document_path: str
    
    start_page: int
    end_page: int
    
    text: str
```

```python
@dataclass
class PositionedWord:
    word: str
    page_number: int
```

```python
class FixedSizeChunker:

    def __init__(self, chunk_size=150, overlap=50):
        
        self.chunk_size = chunk_size
        self.overlap = overlap
        self.step = chunk_size - overlap
        

    def get_chunks(self, processed_documents):
        processed_chunks = []

        document_id = 1
        for processed_document in processed_documents:
            positioned_words = []
            
            for page in processed_document.pages:
                words = page.text.split()
            
                for word in words:
                    positioned_words.append(
                        PositionedWord(
                            word = word,
                            page_number = page.page_number
                        )
                    )

            start = 0
            chunk_id = 1

            chunks = []
            while start < len(positioned_words):
    
                chunk_words = positioned_words[start : start + self.chunk_size]
                chunk_text = " ".join(word.word for word in chunk_words)
            
                page_numbers = [word.page_number for word in chunk_words]
                
                chunks.append(
                    Chunk(
                        chunk_id = chunk_id,

                        document_id = document_id,
                        document_path = processed_document.file_path,
    
                        start_page = page_numbers[0],
                        end_page = page_numbers[-1],
    
                        text = chunk_text
                    )
                )
    
                start = start + self.step
                chunk_id += 1

            processed_chunks.extend(chunks)
            document_id += 1

        return processed_chunks
```

```python
@dataclass
class EmbeddedChunk:
    chunk: Chunk
    embedding: np.ndarray
```

```python
class EmbeddingService:

    def __init__(self):

        self.embedding_model = SentenceTransformer(
            "../models/embeddings/bge-base-en-v1.5"
        )

    def embed_documents(self, chunks):

        texts = [chunk.text for chunk in chunks]

        embeddings = self.embedding_model.encode(
            texts,
            convert_to_numpy=True,
            show_progress_bar=True
        )

        return embeddings
    

    def embed_query(self, query):

        query = (
            "Represent this sentence for searching relevant passages: "
            + query
        )

        embedding = self.embedding_model.encode(
            query,
            convert_to_numpy=True
        )

        return embedding
```

```python
class FaissStore:
    
    def __init__(self, dim=768, use_existing_data=False, embeddings_paths={}):

        self.index_file_path = embeddings_paths["index_file_path"]
        self.chunks_file_path = embeddings_paths["chunks_file_path"]
        self.knowledge_base_path = embeddings_paths["knowledge_base_path"]

        self.index = None
        self.chunk_metadata = None
        self.chunks = None
        
        if not use_existing_data:
            for path in [self.index_file_path, self.chunks_file_path, self.knowledge_base_path]:
                if os.path.exists(path):
                    os.remove(path)
                    
            self.index = faiss.IndexFlatIP(dim)
            self.chunk_metadata = []
            self.chunks = {}

        else:
            self.index = faiss.read_index(self.index_file_path)
            with open(self.knowledge_base_path, "rb") as file:
                self.chunk_metadata = list(pickle.load(file))

            with open(self.chunks_file_path, "rb") as file:
                self.chunks = dict(pickle.load(file))

    
    def add_chunks(self, chunks, embeddings):
        embeddings = embeddings.astype(np.float32)
        faiss.normalize_L2(embeddings)

        self.index.add(embeddings)

        for chunk_info in chunks:
            self.chunk_metadata.append(
                (chunk_info.document_id, chunk_info.chunk_id)
            )

        self.save(chunks)


    def save(self, chunks):
        faiss.write_index(self.index, self.index_file_path)
        with open(self.knowledge_base_path, 'wb') as file:
            pickle.dump(self.chunk_metadata, file)

        updated_chunks = {}
        for chunk in chunks:
            updated_chunks.update({
                (chunk.document_id, chunk.chunk_id): {
                    "document_path": chunk.document_path, 
                    "start_page": chunk.start_page, 
                    "end_page": chunk.end_page, 
                    "text": chunk.text
                }
            })

        self.chunks.update(updated_chunks)
        with open(self.chunks_file_path, 'wb') as file:
            pickle.dump(self.chunks, file)

    
    def search(self, query_embedding, top_k=5):
        
        query_embedding = query_embedding.reshape(1, -1).astype(np.float32)
        faiss.normalize_L2(query_embedding)

        scores, indices = self.index.search(query_embedding, top_k)

        results = []
        for score, idx in zip(scores[0], indices[0]):

            if idx == -1:
                continue

            results.append({
                "score": float(score),
                "chunk": self.chunk_metadata[idx]
                })

        return results
```

```python
class Retriever:

    def __init__(self, embedding_service, vector_store):
        self.embedding_service = embedding_service
        self.vector_store = vector_store


    def retrive(self, query, top_k=10):
        query_embedding = self.embedding_service.embed_query(query)

        results = self.vector_store.search(
            query_embedding,
            top_k
        )

        final_results = []
        for result in results:

            final_results.append(
                self.vector_store.chunks[result["chunk"]]
            )
            
        return final_results
```

```python
class ContextBuilder:

    def build_context(self, chunks):

        context = "\n\n".join(
            chunk["text"]
            for chunk in chunks
        )
        
        return context
```

```python
class LLM_Service:

    def __init__(self, model):
        self.model = model
        
    def generate(self, prompt):
        stream = ollama.chat(
            model=self.model,
            messages=[
                {
                    "role": "user",
                    "content": prompt
                }
            ],
            stream=True
        )
        
        for chunk in stream:
            print(
                chunk["message"]["content"],
                end="",
                flush=True
            ) 
```

```python
class RAG:
    
    def __init__(self, config=None):
        if not config:
            print("Please provide configurations.")

        if not config["use_existing_data"]:
            pdf_loader = PDFLoader()
            processed_documents = pdf_loader.load_documents(config["documents"])

            chunker = FixedSizeChunker()
            chunks = chunker.get_chunks(processed_documents)
        
        embedding_service = EmbeddingService()
        if not config["use_existing_data"]:
            embeddings = embedding_service.embed_documents(chunks)

        vector_store = FaissStore(use_existing_data=config["use_existing_data"], embeddings_paths=config["embeddings_data"])
        if not config["use_existing_data"]:
            vector_store.add_chunks(chunks, embeddings)

        
        self.retriever = Retriever(embedding_service, vector_store)
        self.context_builder = ContextBuilder()

        self.llm_model = LLM_Service(config["llm_model"])


    def build_prompt(self, query, context):
        prompt = f"""You are a helpful AI assistant.
        
        Answer the user's question using only the provided context.
        
        Requirements:
        1. Give a concise answer.
        2. Focus only on information relevant to the question.
        3. Do not include unrelated details.
        4. If the answer is not in the context, say so.

        When writing equations:
        1. Do not use LaTeX.
        2. Use plain text mathematical notation.
        
        Context:
        {context}
        
        Question:
        {query}
        
        Answer:
        """

        return prompt

    def ask(self, query):
        result = self.retriever.retrive(query)

        for res in result:
            print(
                res["start_page"],
                res["end_page"],
                res["text"][:100]
            )
        context = self.context_builder.build_context(result)

        print("\nCONTEXT : \n", context, "\n")
        prompt = self.build_prompt(query, context)
        
        self.llm_model.generate(prompt)
```

```python
def score_answer(answer, expected_keywords):

    answer = answer.lower()

    matched_keywords = []
    for keyword in expected_keywords:

        if keyword.lower() in answer:
            matched_keywords.append(keyword)

    score = len(matched_keywords) / len(expected_keywords)

    return {
        "score": round(score, 2),
        "matched_keywords": matched_keywords,
        "missing_keywords": [
            kw for kw in expected_keywords
            if kw not in matched_keywords
        ]
    }
```

```python
avg_score = 0

for sample in evaluation_dataset:

    results = rag.evaluate_retrieval_with_bm25(sample["question"])
    
    score = score_answer(
        results["context"],
        sample["expected_keywords"]
    )

    print(
        sample["question"],
        score, 
        "\n"
    )

    avg_score += score["score"]

print("Final Retrival Score : ", avg_score / len(evaluation_dataset))
```

```python
from rank_bm25 import BM25Okapi

class BM25_Retriever:

    def __init__(self, vector_store):

        self.vector_store = vector_store

        self.chunk_keys = []
        self.tokenized_chunks = []

        for key, chunk in self.vector_store.chunks.items():

            self.chunk_keys.append(key)

            self.tokenized_chunks.append(
                chunk["text"].lower().split()
            )

        self.bm25 = BM25Okapi(
            self.tokenized_chunks
        )


    def retrieve(self, query, top_k=5):
    
        tokenized_query = (
            query.lower().split()
        )
    
        scores = self.bm25.get_scores(tokenized_query)
    
        ranked_indices = sorted(
            range(len(scores)),
            key=lambda i: scores[i],
            reverse=True
        )[:top_k]
    
        results = []
        for idx in ranked_indices:
    
            chunk_key = self.chunk_keys[idx]
            
            results.append(
                self.vector_store.chunks[
                    chunk_key
                ]
            )
    
        return results
```
